### Question 3: Write a Python program to remove the first occurrence of a specified element from a list.

- `sample input : `
    - `var = [1,2,3,4,2,5]`
- `sample output :`
    - `result : [1,3,4,2,5]`
    

```python
# Please write The solution Here for above question 3:
var = [1,2,3,4,2,5]
a=[]
for i in var:
  if i not in a:
    a.append(i)
  else:
    a.remove(i)
    a.append(i)

print("result :",a)



```

### Question 8: Write a Python program to find the common elements between two lists.

- `sample input : `
    - `var1 = [1,2,3,4,10]`
    - `var2 = [10,11,12,14]`
- `sample output :`
    - `result : 10`

```python
# Please write The solution Here for above question 8:
var1 = [1,2,3,4,10]
var2 = [10,11,12,14]
a=[]
for i in var1:
  if i in var2:
    a.append(i)
print("result:",a)


```
